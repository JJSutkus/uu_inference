## Q7


```python
import pandas as pd
import numpy as np

ames = pd.read_csv("./data/ames_prices.csv")
ames["Central.Air"] = ames["Central.Air"].map({"Y": 1, "N": 0})
ames.columns
```

```python
ames["Central.Air"].mean()
```

a. The proportion of homes in this dataset with central air appears to be around 93.3%. 

```python
# b.

import random
random.seed(123)

means = []
B = 2000

for i in range(0,B):
    sample = ames.sample(n=100, replace=True)
    means.append(sample["Central.Air"].mean())

sum(means)/len(means)
```

```python
# c.

std_error = np.std(means, ddof=1)
std_error
```

```python
# d.

p_hat = ames["Central.Air"].mean()

critical_values = {
    "90%": 1.645,
    "95%": 1.960,
    "99%": 2.576
}

for level, z in critical_values.items():
    lower = p_hat - z * std_error
    upper = p_hat + z * std_error
    print(f"{level} confidence interval: ({lower:.4f}, {upper:.4f})")
```

e. For null hypothesis Ho: p = .9, we fail to reject the null hypothesis at .1, .05, and .01 significance levels since .9 falls within the confidence interval for the 90%, 95% and 99% confidence interval. For interpretation, we are 95% confident the true proprotion of home with central air in ames is between roughly .8835 and .9827.

## Q8

```python
# a.

random.seed(456)

p_hat = np.arcsinh(ames["area"]).corr(np.arcsinh(ames["price"]))

B = 2000
corrs = []

for i in range(0,B):
    sample = ames.sample(n=100, replace=True)
    corr = np.arcsinh(sample["area"]).corr(np.arcsinh(sample["price"]))
    corrs.append(corr)

sum(corrs)/len(corrs)
```

```python
# b.

std_error2 = np.std(corrs, ddof=1)
std_error2
```

```python
# c.

critical_values = {
    "90%": 1.645,
    "95%": 1.960,
    "99%": 2.576
}

for level, z in critical_values.items():
    lower = p_hat - z * std_error2
    upper = p_hat + z * std_error2
    print(f"{level} confidence interval: ({lower:.4f}, {upper:.4f})")
```

d. For significance levels .1, .05, and .01, we will reject the null hypothesis Ho: p=0 as 0 is excluded from the 90%, 95%, and 99% confidence interval suggesting that there is sufficient evidence to conclude that there is a noticable correlation between area and price of a home. 

e. Area and price must remain paired during resampling since each observation represents one home's area and its corresponding price. Resampling the two variables independently would break these pairings and distort the relationship, producing an inaccurate estimate of the correlation.

## Q9

```python
meta = pd.read_csv("./data/metabric.csv")
meta.head()
```

```python
# a.

random.seed(789)

x_bar = meta["Overall Survival (Months)"].mean()

means2 = []
B = 2000

for i in range(0,B):
    sample = meta.sample(n=100, replace=True)
    means2.append(sample["Overall Survival (Months)"].mean())

sum(means2)/len(means2)
```

```python
# b.

std_error3 = np.std(means2, ddof=1)
std_error3
```

```python
# c.

critical_values = {
    "90%": 1.645,
    "95%": 1.960,
    "99%": 2.576
}

for level, z in critical_values.items():
    lower = x_bar - z * std_error3
    upper = x_bar + z * std_error3
    print(f"{level} confidence interval: ({lower:.4f}, {upper:.4f})")
```

d. For significance levels .1, .05, and .01, we will fail to reject the null hypothesis Ho: m=120 as 120 is included from the 90%, 95%, and 99% confidence interval suggesting that there is not sufficient evidence to conclude that mean overall survival months is not 120. 

e. 
- The estimated mean surival time is about 129.0943 months, or about 10.76 years.
- 90% confidence interval: average survival is 116.3841–141.8045 months, or approximately 9.7–11.82 years
- 95% confidence interval: average survival is 113.9502–144.2384 months, or approximately 9.5–12.02 years
- 99% confidence interval: average survival is 109.1906–148.9979 months, or approximately 9.1–12.42 years

Because some patients are still alive at the end of the study, their eventual death times are right-censored (we know they survived at least until their last observation, but do not represent their total lifetimes). Treating these observed times as completed lifetimes would underestimate mean survival time. Therefore, these estimates do not necessarily represent the true population mean survival time.

```python
nhanes = pd.read_csv("./data/nhanes_data_17_18.csv")
nhanes = nhanes[nhanes["AgeInYearsAtScreening"]>18]
nhanes = nhanes[nhanes["SleepHoursWeekdaysOrWorkdays"].notnull()]
nhanes.head()
```

```python
# a.

random.seed(123456789)

x_bar2 = nhanes["SleepHoursWeekdaysOrWorkdays"].mean()

B = 2000
means3 = []

for i in range(0,B):
    sample = nhanes.sample(n=100, replace=True)
    means3.append(sample["SleepHoursWeekdaysOrWorkdays"].mean())

sum(means3)/len(means3)
```

```python
# b.

std_error4 = np.std(means3, ddof=1)
std_error4
```

```python
# c.

critical_values = {
    "90%": 1.645,
    "95%": 1.960,
    "99%": 2.576
}

for level, z in critical_values.items():
    lower = x_bar2 - z * std_error4
    upper = x_bar2 + z * std_error4
    print(f"{level} confidence interval: ({lower:.4f}, {upper:.4f})")
```

d. At significance levels .1 and .05, we reject the null hypothesis Ho: m=8 (hours of sleep) since the 90% and 95% confidence intervals do not include 8, but at significance level .01, we fail to reject the null hypothesis since 8 is contained in the 99% confidence interval.

e. The estimated mean amount of sleep is around 7.6055 hours, which is about 24 minutes less than the null hypothesis 8 hours. Statistical and practical significance do not necessarily tell the same story. Although the difference of approximately 24 minutes is statistically significant at the .1 and .05 levels, whether it is practically important depends on the context, such as the effects of this difference on health, concentration, daily functioning, etc.. A difference of 24 minutes may be meaningful in some cases, but statistical significance alone does not establish that the difference has meaningful impact in the real world.
